```python
import subprocess
import os
```

```python
# Base directory
base_directory = "/media/toor/T7Shield/Mouse1637/Round3/TMaze"
```

```python
# Directory to change to before running the bash script
bash_working_directory = "/home/toor/Desktop/stability-preprocessing/"
os.chdir(bash_working_directory)
python_env = "/home/toor/miniconda3/envs/stability-preprocessing-keras/bin/python"
```

```python
# Walk through the directory and subdirectories
for root, dirs, files in os.walk(base_directory):
    # Initialize file paths
    miniscope_file = None
    motion_correction = None

    # Look for the required files in the current directory
    for file in files:
        if "miniscope" in file:
            miniscope_file = os.path.join(root, file)
            
        if "motion_correction_comparison_final2" in file: # this means that the file is already processed
            motion_correction = os.path.join(root, file)

        # Check if all required files are found
    if miniscope_file and not motion_correction:
        # use os.path.expanduser() to handle the home directory
        script_path = os.path.expanduser("preproc_caiman_final.py")
        
        # Construct and run the bash command
        bash_command = (
                    f"conda run -n stability-preprocessing-keras "
                    f"python {script_path} "
                    f"--input_path \"{miniscope_file}\""
                )

        print("Running:", bash_command)

        try:
            subprocess.run(bash_command, shell=True, check=True)
        except subprocess.CalledProcessError as e:
            print(f"Error while processing {root}: {e}")
```
